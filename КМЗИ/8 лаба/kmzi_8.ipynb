```python
#Разложение числа на эллиптической кривой

def number_n():
    while(1):
        n=randint(2**10, 2**15)
        print("n = ",n)
        #n=randint(2**40, 2**45)
        if(is_prime(n)==0 and n % 2 != 0):## можно с делением на 2 поэксперементировать
            return n
#сначал сгенерируем кривую

def curve_generation(n):
    Q_x = randint(2, n-1)
    Q_y = randint(2, n-1)
    A = randint(1, n)
    B = (Q_y**2 - Q_x**3 - A*Q_x)%n
    F = Zmod(n)
    E = EllipticCurve(F, [A, B])
    Q_0=E.point((Q_x, Q_y))
    elements = [F(i) for i in range(n)]
    return Q_0, A, B, F, E,elements


#n = number_n()
n = 10403
```

```python
# сложение точек эллиптической кривой
def Sum_point(x_1, y_1, x_2, y_2, A, n):
    if x_1 == x_2 and y_1 == y_2:
        try:
            denominator = (2 * y_1) % n
            lamda = ((3 * x_1**2 + A) / (denominator)) % n
        except ZeroDivisionError:
            if (2 * y_2) % n == 0:
                return 1, 0, 1
            else:
                denominator = (2 * y_1) % n
                return 'Error', 'Error', denominator
    else:
        try:
            denominator = (x_2 - x_1) % n
            lamda = ((y_2 - y_1) / (denominator)) % n
        except ZeroDivisionError:
            if (x_2 - x_1) % n == 0:
                return 1, 0, 1
            denominator = (x_2 - x_1) % n
            return 'Error', 'Error', denominator
    res_x = (lamda**2 - x_1 - x_2) % n
    res_y = (lamda * (x_1 - res_x) - y_1) % n
    return res_x, res_y, denominator
```

```python
import math
import time

def factorization_number(n,m):
    #Выбрать случайную эллиптическую кривую 𝐸(ℤ𝑛ℤ⁄) и точ-ку 𝑄 на ней.
    i = 2
    while 1:
        Q_0, A, B, F, E, elements = curve_generation(n)    
        m_max=(len(elements)) # база разложения, начтаь с i = 2
        if  m > m_max:
            continue
        Q_i=Q_0
        while i < m:
            p_i=elements[i]
            i = i + 1
            alfa_i = int(1/2 * (log(n)/log(int(p_i))))
            j=0
            Q_x = int(Q_i[0])
            Q_y = int(Q_i[1])
            while j < alfa_i:
                if p_i==2:
                    lamda_list=[]
                    Q_x, Q_y, lamda = Sum_point(Q_x, Q_y, Q_x, Q_y, A, n)
                    lamda_list.append(lamda)
                else:
                    lamda_list = []
                    Q_x_2, Q_y_2 = Q_x, Q_y
                    for i in range(1, p_i):
                        Q_x, Q_y, lamda = Sum_point(Q_x, Q_y, Q_x_2, Q_y_2, A, n)
                        lamda_list.append(lamda)

                gcd_list = []
                d = []
                for lamda_new in lamda_list:
                    gcd_list.append(gcd(n, lamda_new))
                for gcd_new in gcd_list:
                    if gcd_new > 1 and gcd_new < n:
                        d.append(gcd_new)
                if d != []:
                    return Q_0, A, B, F, E,elements,d[0]
                j += 1
    print(f"Количество итераций = {i}")
    return Q_0, A, B, F, E,elements

start = time.time()
Q_0, A, B, F, E, elements, d = factorization_number(n,10)   
finish = time.time()
res = finish - start
print(f"Время работы программы = {res * 1000} миллисекунд")
print(f"n = {n}")
print(f"Коэффициенты кривой: A = {A}, B = {B}")
print(f"Точка Q = {Q_0}")
print(f"d = {d}")
q=n/d
print(f"q = n/d = {q}")
```
